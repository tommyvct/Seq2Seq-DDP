```python
corpus = ['stac', 'molweni']
struct_type = ['focus', 'natural2', 'augmented', 'natural']
t5_family = [ 'flan-t5', 't5', 't5gemma2'] # 't0-3b',
model_size = {
    'flan-t5': ['base', 'large'],
    't5': ['large'],
    't5gemma2': ['270m']
    # 't0-3b': [''],
    # 'flan-t5': ['base', 'large', 'xl'],
    # 't5': ['3b', 'large'],
    # 't5gemma2': ['270m', '1b', '4b']
}
seed = 27
lr = "5e-5"
```

```python
for c in corpus:
    for s in struct_type:
        for t in t5_family:
            for m in model_size[t]:
                if s == 'focus' or s == 'natural2':
                    script = "transition_predict.py"
                else:
                    script = "train.py"
                msz = "" if t == 't0-3b' else f"-m {m}"
                full_cmd = f"python3 {script} --train_corpus {c} --test_corpus {c} -s {s} -t {t} {msz} --lr {lr} --seed {seed} -b"
                print(full_cmd)
```
